# IPL Prediction Numpy

## Task 0 (Data Loading)

```python
import numpy as np

file_path = 'data/deliveries.csv'
header = np.genfromtxt(file_path, delimiter=',', dtype=str, max_rows=1)
data = np.genfromtxt(file_path, delimiter=',', dtype=str, skip_header=1)

match_id_idx = np.where(header == 'match_id')[0][0]
batting_team_idx = np.where(header == 'batting_team')[0][0]
batter_idx = np.where(header == 'batter')[0][0]
bowler_idx = np.where(header == 'bowler')[0][0]
batsman_runs_idx = np.where(header == 'batsman_runs')[0][0]
over_idx = np.where(header == 'over')[0][0]
extras_idx = np.where(header == 'extra_runs')[0][0]

match_ids_str = data[:, match_id_idx]
batting_team = data[:, batting_team_idx]
batter = data[:, batter_idx]
bowler = data[:, bowler_idx]
batsman_runs_str = data[:, batsman_runs_idx]
over_str = data[:, over_idx]
extras_str = data[:, extras_idx]

match_ids_str = np.where(match_ids_str == '', '0', match_ids_str)
batsman_runs_str = np.where(batsman_runs_str == '', '0', batsman_runs_str)
over_str = np.where(over_str == '', '0', over_str)
extras_str = np.where(extras_str == '', '0', extras_str)

match_ids = match_ids_str.astype(int)
batsman_runs = batsman_runs_str.astype(int)
over = over_str.astype(int)
extras = extras_str.astype(int)

print("Task 0: Data loaded successfully!")
```

## Task 1 (TOTAL RUNS PER MATCH)

```python
unique_matches = np.unique(match_ids)

total_runs_masked = [
    (int(match), int(np.sum(batsman_runs[match_ids == match]) + np.sum(extras[match_ids == match])))
    for match in unique_matches
]

print("\nTask 1 (Boolean Masking) - First 5 matches:")
print(total_runs_masked[:5])
```

## Task 2 (Top 5 Batters)

```python
uniq_batters, inverse_indices = np.unique(batter, return_inverse=True)

batter_totals = np.bincount(inverse_indices, weights=batsman_runs).astype(int)

sorted_indices = np.argsort(batter_totals)

top_5_indices = sorted_indices[-5:][::-1]

top_5_batters = uniq_batters[top_5_indices]
top_5_runs = batter_totals[top_5_indices]

top_5_result = list(zip(top_5_batters.tolist(), top_5_runs.tolist()))

print("\nTask 2: Top 5 Batters")
for player in top_5_result:
    print(player)
```

## Task 3 (Strike Rates)

```python
uniq_batters, inverse_indices = np.unique(batter, return_inverse=True)

batter_runs = np.bincount(inverse_indices, weights=batsman_runs)

balls_faced = np.bincount(inverse_indices)

strike_rates = (batter_runs / balls_faced) * 100

strike_rates = np.round(strike_rates, 2)

strike_rate_result = list(zip(uniq_batters.tolist(), strike_rates.tolist()))

print("\nTask 3: Sample Strike Rates (First 5 batters alphabetically)")
for player in strike_rate_result[:5]:
    print(player)
```

## Task 4 (Economy Rates)

```python

uniq_bowlers, inv_indices = np.unique(bowler, return_inverse=True)

runs_conceded = np.bincount(inv_indices, weights=batsman_runs)

balls_bowled = np.bincount(inv_indices)

overs_bowled = balls_bowled / 6

economy_rates = runs_conceded / overs_bowled

economy_rates = np.round(economy_rates, 2)

economy_result = list(zip(uniq_bowlers.tolist(), economy_rates.tolist()))

print("\nTask 4: Sample Economy Rates (First 5 bowlers alphabetically)")
for player in economy_result[:5]:
    print(player)
```

## Task 5 (Average Runs per Over)

```python
uniq_overs, inv_overs = np.unique(over, return_inverse=True)

runs_in_over = np.bincount(inv_overs, weights=batsman_runs)

balls_in_over = np.bincount(inv_overs)

avg_runs_per_over = (runs_in_over / balls_in_over) * 6

avg_runs_per_over = np.round(avg_runs_per_over, 2)

avg_runs_list = avg_runs_per_over.tolist()

print("\nTask 5: Average Runs per Over")
for o, avg in zip(uniq_overs, avg_runs_list):
    print(f"Over {o}: {avg}")

print("\nRaw Output List:")
print(avg_runs_list)
```

## Task 6 (Boundary Analysis)

```python

total_fours = np.sum(batsman_runs == 4)
total_sixes = np.sum(batsman_runs == 6)

print("\nTask 6: Boundary Counts")
print(f"Total Fours: {total_fours}")
print(f"Total Sixes: {total_sixes}")

is_boundary = (batsman_runs == 4) | (batsman_runs == 6)

teams_hitting_boundaries = batting_team[is_boundary]

uniq_teams, boundary_counts = np.unique(teams_hitting_boundaries, return_counts=True)

top_team_idx = np.argmax(boundary_counts)

top_boundary_team = uniq_teams[top_team_idx]
max_boundaries = boundary_counts[top_team_idx]

print("\nBonus: Team with Most Boundaries")
print(f"{top_boundary_team} dominates with {max_boundaries} boundaries!")
```

## Task 7 (Death Overs Analysis)

```python

death_overs_mask = (over >= 16) & (over <= 20)

total_death_runs = np.sum(batsman_runs[death_overs_mask])

print("\nTask 7: Death Overs Analysis")
print(f"Total runs scored in all death overs: {total_death_runs}")


death_teams = batting_team[death_overs_mask]
death_runs_array = batsman_runs[death_overs_mask]

uniq_death_teams, death_inv_indices = np.unique(death_teams, return_inverse=True)

team_death_totals = np.bincount(death_inv_indices, weights=death_runs_array).astype(int)

top_team_idx = np.argmax(team_death_totals)

top_death_team = uniq_death_teams[top_team_idx]
max_death_runs = team_death_totals[top_team_idx]

print(f"Most Dangerous Death Over Team: {top_death_team} with {max_death_runs} runs!")
```

## Task 8 (Highest Scoring Match)

```python

uniq_matches, inv_indices = np.unique(match_ids, return_inverse=True)

match_totals = np.bincount(inv_indices, weights=batsman_runs).astype(int)

highest_match_idx = np.argmax(match_totals)

best_match_id = uniq_matches[highest_match_idx]
highest_runs = match_totals[highest_match_idx]

highest_scoring_match = (int(best_match_id), int(highest_runs))

print("\nTask 8: Highest Scoring Match")
print(highest_scoring_match)
```

## Task 9 (Runs Per Team Per Match)

```python

match_ids_str_array = match_ids.astype(str)

underscore_array = np.array(['_'] * len(match_ids))

composite_keys = np.char.add(match_ids_str_array, underscore_array)
composite_keys = np.char.add(composite_keys, batting_team)

uniq_combos, inv_combos = np.unique(composite_keys, return_inverse=True)

combo_runs = np.bincount(inv_combos, weights=batsman_runs).astype(int)

print("\nTask 9: Runs per Team per Match (Sample of 10 innings)")
print("-" * 45)
print(f"{'MATCH ID':<10} | {'TEAM':<30} | TOTAL RUNS")
print("-" * 45)

for combo, runs in zip(uniq_combos[:10], combo_runs[:10]):
    match_id, team = combo.split('_')
    print(f"{match_id:<10} | {team:<30} | {runs}")
```

## Task 10 (Match Winner Approximation)

```python
underscore_array = np.array(['_'] * len(match_ids))
match_ids_str_array = match_ids.astype(str)

composite_keys = np.char.add(match_ids_str_array, underscore_array)
composite_keys = np.char.add(composite_keys, batting_team)
uniq_combos, inv_combos = np.unique(composite_keys, return_inverse=True)
combo_runs = np.bincount(inv_combos, weights=batsman_runs).astype(int)

print("\nTask 9: Match Winner Approximation (Sample of 5 Matches)")

sample_matches = np.unique(match_ids)[:5]

for match in sample_matches:
    match_prefix = str(match) + "_"
    mask = np.char.startswith(uniq_combos, match_prefix)
    match_teams_combos = uniq_combos[mask]
    match_teams_runs = combo_runs[mask]

    if len(match_teams_runs) == 2:
        team1_runs, team2_runs = match_teams_runs
        team1_name = match_teams_combos[0].split('_')[1]
        team2_name = match_teams_combos[1].split('_')[1]

        if team1_runs > team2_runs:
            winner = team1_name
        elif team2_runs > team1_runs:
            winner = team2_name
        else:
            winner = "Tie"

        print(f"Match {match}: {team1_name} ({team1_runs}) vs {team2_name} ({team2_runs}) -> Winner: {winner}")
    else:
        print(f"Match {match}: Abnormal innings count (Rain or Super Over).")
```

## Task 11 (Match Scorecard Generation)

```python
match_ids_str_array = match_ids.astype(str)
underscore_array = np.array(['_'] * len(match_ids))
composite_keys = np.char.add(match_ids_str_array, underscore_array)
composite_keys = np.char.add(composite_keys, batting_team)
uniq_combos, inv_combos = np.unique(composite_keys, return_inverse=True)
combo_runs = np.bincount(inv_combos, weights=batsman_runs).astype(int)
uniq_matches = np.unique(match_ids)
sorted_matches = np.sort(uniq_matches)

print("\nTask 11: Match Scorecard Generation (Showing First 5 Matches)")
print("=" * 40)

for match in sorted_matches[:5]:
    match_prefix = str(match) + "_"
    mask = np.char.startswith(uniq_combos, match_prefix)
    match_teams_combos = uniq_combos[mask]
    match_teams_runs = combo_runs[mask]

    print(f"Match {match}:")
    for combo, runs in zip(match_teams_combos, match_teams_runs):
        team_name = combo.split('_')[1]
        print(f" Team {team_name}: {runs} runs")

    print("-" * 30)
```
